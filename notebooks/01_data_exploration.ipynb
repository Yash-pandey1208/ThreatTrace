```python

from collections import Counter

file_reports = []

for file in csv_files:
    # Read only the CSV header first
    header = pd.read_csv(file, nrows=0).columns
    normalized_columns = header.str.strip()

    label_matches = [
        col for col in header
        if col.strip().lower() == "label"
    ]

    if len(label_matches) != 1:
        print(f"Check label column manually: {file.name}")
        continue

    label_col = label_matches[0]
    label_counts = Counter()
    total_rows = 0

    # Read only the label column in chunks
    for chunk in pd.read_csv(
        file,
        usecols=[label_col],
        chunksize=100_000,
        low_memory=False
    ):
        labels = chunk[label_col].astype("string").str.strip()
        label_counts.update(labels.fillna("<MISSING>").tolist())
        total_rows += len(chunk)

    file_reports.append({
        "file": file.name,
        "rows": total_rows,
        "columns": len(normalized_columns),
        "labels": dict(label_counts)
    })

    print(f"\n{file.name}")
    print("Rows:", total_rows)
    print("Columns:", len(normalized_columns))
    print("Labels:", dict(label_counts))
```
